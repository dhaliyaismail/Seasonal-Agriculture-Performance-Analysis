## Univariate: Season Distribution(Pie Chart)

```python
plt.figure(figsize=(7,7))
df['Season'].value_counts().plot.pie(autopct='%1.1f%%',startangle=90,cmap='viridis')
plt.title("Distribution of Records by Season")
plt.ylabel("") #Default 'Season' label hidden in the y-axis
plt.tight_layout()
plt.show
```

```python
#IQR-based outlier summary for numerical columns
outlier_summary=[]
for col in numeric_columns:
  q1=df[col].quantile(0.25)
  q3=df[col].quantile(0.75)
  iqr=q3-q1
  lower=q1-1.5*iqr
  upper=q3+1.5*iqr
  count=((df[col]<lower)|(df[col]>upper)).sum()
  outlier_summary.append({
      "Column":col,
      "Lower Bound":lower,
      "Upper Bound":upper,
      "Outlier Count":count,
      "Outlier Percentage":round(count/len(df)*100,2)
  })
outlier_summary=pd.DataFrame(outlier_summary).sort_values(
    "Outlier Count",ascending=False)
outlier_summary
```

## 13.Additional Student Driven Analysis


### Student Analysis 1(Resource Usage)
Water is a very essential input in farming. Measuring yield per unit of water shows which season uses water most efficiently.

```python
# Water Use Efficiency = Yield per unit Water

df["Water_Efficiency"] = df["Yield_Tonnes_Ha"] / df["Water_Used_m3"]
efficiency_summary = df.groupby("Season")["Water_Efficiency"].mean().sort_values(ascending=False)
print(efficiency_summary)
plt.figure(figsize=(8,5))
sns.barplot(x=efficiency_summary.index, y=efficiency_summary.values, palette="crest")
plt.title("Water Use Efficiency (Yield per m³ Water) by Season")
plt.xlabel("Season")
plt.ylabel("Yield per Unit Water (Tonnes/Ha per m³)")
plt.tight_layout()
plt.show()
```

### Student Analysis 2(Environmental Conditions)
Temperature is a key factor affecting crop growth.Whether yield tends to rise or fall as temperature changes is analysed.

```python
# Effect of Temperature on Yield

plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x="Avg_Temperature_C", y="Yield_Tonnes_Ha", hue="Season", alpha=0.6)
plt.title("Effect of Temperature on Agricultural Yield")
plt.xlabel("Average Temperature (°C)")
plt.ylabel("Yield (Tonnes/Ha)")
plt.legend(title="Season")
plt.tight_layout()
plt.show()
```

### Student Analysis 3(Economic Performance)
Farms vary in size. Dividing profit by farm area gives a fair measure of true economic performance across different  seasons.

```python
# Additional Analysis 3 (Direction: Economic Performance)
df["Profit_per_Ha"] = df["Profit_INR"] / df["Farm_Area_Hectares"]
profit_ha_summary = df.groupby("Season")["Profit_per_Ha"].mean().sort_values(ascending=False)
print(profit_ha_summary)
plt.figure(figsize=(8,5))
sns.boxplot(data=df, x="Season", y="Profit_per_Ha")
plt.title("Profit per Hectare across Seasons")
plt.xlabel("Season")
plt.ylabel("Profit per Hectare (INR/Ha)")
plt.tight_layout()
plt.show()
```

## 16.Conclusion
This analysis examined how agricultural performance varies across seasons, looking at yield, profit, water usage, rainfall, temperature, irrigation method, and crop type. One season stood out with consistently higher average yield and profit per hectare, while another showed greater variability, suggesting less predictable outcomes for farmers planting during that period. Rainfall and temperature both showed some association with yield, though the relationships were moderate rather than strong.
The most useful findings came from the season-level summary statistics, the water-use efficiency comparison, and the crop-by-season breakdown, which together showed that certain crops and irrigation methods performed more consistently well than others.
These results should be read as associations, not proven causes — the dataset doesn't capture factors like soil quality, pest pressure, or market prices, and some crop-season groups had limited records, which affects reliability. Even so, the analysis offers a useful starting point for identifying which seasons, crops, and practices deserve closer investigation, and can support preliminary planning decisions while further data collection helps confirm these patterns.
